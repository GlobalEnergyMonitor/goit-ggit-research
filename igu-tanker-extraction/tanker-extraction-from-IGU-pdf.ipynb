```python
import pdfplumber
import pandas as pd
```

```python
# Path to your PDF file
pdf_path = "/Users/baird/Downloads/IGU2024-LNG-Report_Final_LR.pdf"

# List to hold DataFrames for all extracted tables
tables_list = []

# Open the PDF file
with pdfplumber.open(pdf_path) as pdf:
    # pdf.pages is 0-indexed; pages 122 to 152 correspond to indices 121 to 151
    #for i in range(121, 152):
    for i in range(60, 77):
        page = pdf.pages[i]
        # Extract all tables from the current page
        page_tables = page.extract_tables()
        # For each table, if it has more than one row, convert it to a DataFrame
        for table in page_tables:
            if len(table) > 1:
                df = pd.DataFrame(table[1:], columns=table[0])
                tables_list.append(df)

# Now, print the CSV-formatted text for each extracted table
for j, df in enumerate(tables_list, start=1):
    csv_text = df.to_csv(index=False)
    print(f"==== Table {j} (extracted from pages 122-152) ====")
    print(csv_text)
    print("\n")
```
